```python
import pandas as pd

# Example
# requests = [
#     {"queries":["q1","q2"], "keys":["k1","k2"]},
#     {"queries":["q3"], "keys":["k3","k4","k5"]}
# ]

# Step 1: Build DataFrame
df = pd.DataFrame(requests)

# Step 2: Explode queries, then explode keys (order matters for IDs)
df = df.reset_index(names="request_id") \
       .explode("queries") \
       .explode("keys")

# Step 3: Assign query_id (position within original queries list per request)
df["query_id"] = df.groupby("request_id")["queries"].cumcount()

# Step 4: Assign key_id (position within original keys list per request)
# To do this correctly, regroup by request_id and "queries"
df["key_id"] = df.groupby(["request_id", "queries"])["keys"].cumcount()

# Step 5: Reorder and rename
final_df = df.rename(columns={"queries": "query", "keys": "key"})[
    ["request_id", "query_id", "key_id", "query", "key"]
]

print(final_df)
```

```python
import pandas as pd

# Example
# requests = [
#     {"queries":["q1","q2"], "keys":["k1","k2"]},
#     {"queries":["q3"], "keys":["k3","k4","k5"]}
# ]

all_dfs = []

for rid, req in enumerate(requests):
    queries = pd.DataFrame({"query": req["queries"]})
    queries["query_id"] = queries.index

    keys = pd.DataFrame({"key": req["keys"]})
    keys["key_id"] = keys.index

    # Cartesian product (cross join)
    cross = queries.merge(keys, how="cross")
    cross["request_id"] = rid

    all_dfs.append(cross)

final_df = pd.concat(all_dfs, ignore_index=True)[
    ["request_id", "query_id", "key_id", "query", "key"]
]

print(final_df)
```

```python
import pandas as pd

# Example
# requests = [
#     {"queries":["q1","q2"], "keys":["k1","k2"]},
#     {"queries":["q3"], "keys":["k3","k4","k5"]}
# ]

# Step 1. Put into DataFrame
df = pd.DataFrame(requests).reset_index(names="request_id")

# Step 2. Explode queries with position -> query table
qdf = df[["request_id", "queries"]].explode("queries").reset_index(drop=True)
qdf["query_id"] = qdf.groupby("request_id").cumcount()
qdf = qdf.rename(columns={"queries": "query"})

# Step 3. Explode keys with position -> key table
kdf = df[["request_id", "keys"]].explode("keys").reset_index(drop=True)
kdf["key_id"] = kdf.groupby("request_id").cumcount()
kdf = kdf.rename(columns={"keys": "key"})

# Step 4. Cross join queries × keys within each request
final_df = qdf.merge(kdf, on="request_id", how="cross")

print(final_df)
```

```python
import pandas as pd

# Example data
# requests = [
#     {"queries":["q1","q2"], "keys":["k1","k2"]},
#     {"queries":["q3"], "keys":["k3","k4","k5"]}
# ]

# Step 1. Put into DataFrame
df = pd.DataFrame(requests).reset_index(names="request_id")

# Step 2. Explode queries
qdf = df[["request_id", "queries"]].explode("queries").reset_index(drop=True)
qdf["query_id"] = qdf.groupby("request_id").cumcount()
qdf = qdf.rename(columns={"queries": "query"})

# Step 3. Explode keys
kdf = df[["request_id", "keys"]].explode("keys").reset_index(drop=True)
kdf["key_id"] = kdf.groupby("request_id").cumcount()
kdf = kdf.rename(columns={"keys": "key"})

# Step 4. Cross join everything
cross = qdf.merge(kdf, how="cross")

# Step 5. Filter so that only rows with same request_id remain
final_df = cross[cross["request_id_x"] == cross["request_id_y"]] \
    .drop(columns=["request_id_y"]) \
    .rename(columns={"request_id_x": "request_id"})

print(final_df)
```

```python
import pandas as pd

# Example final_df with reward column
final_df = pd.DataFrame({
    "request_id":[0,0,0,1,1,1],
    "query":["q1","q1","q2","q3","q3","q3"],
    "key":["k1","k2","k3","k4","k5","k6"],
    "key_id":[0,1,2,0,1,2],
    "reward":[0.8,0.5,0.9,0.7,0.4,0.6]
})

topn = 2  # choose how many rows per group

# Sort and pick topn per request
df_topn = (
    final_df.sort_values(["request_id", "reward"], ascending=[True, False])
            .groupby("request_id")
            .head(topn)
)

# Collapse into result list per request
result_df = (
    df_topn.groupby("request_id")
    .apply(lambda g: g[["key", "key_id", "reward"]].to_dict("records"))
    .reset_index(name="result")
)

print(result_df)
```
